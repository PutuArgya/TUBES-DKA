## 1. Fuzzification

This section defines:
- The **`trimf`** membership function (from scratch, no skfuzzy)
- The **universe of discourse** for every linguistic variable
- All **membership function sets** for each variable
- A **`fuzzify()`** helper that takes one row of crisp inputs and returns all membership degrees

```python
# ── 1a. Triangular Membership Function ───────────────────────────────────
def trimf(x, abc):
    """
    Triangular membership function (from scratch, no skfuzzy).
    abc : [a, b, c]  –  left foot, peak, right foot.
    Works on both scalar floats and numpy arrays.
    """
    a, b, c = abc
    if a == b:                              # half-triangle: downward slope only
        return np.where(x < b,  0.0,
               np.where(x <= c, (c - x) / (c - b), 0.0))
    if b == c:                              # half-triangle: upward slope only
        return np.where(x < a,  0.0,
               np.where(x <= b, (x - a) / (b - a), 0.0))
    return np.maximum(0, np.minimum((x - a) / (b - a),
                                     (c - x) / (c - b)))
```

```python
# ── 1b. Universe of Discourse ────────────────────────────────────────────
x_global          = np.linspace(0,   5,  500)   # global score  [0 – 5]
x_avg_user_rating = np.linspace(1,   5,  500)   # avg user rating [1 – 5]
x_rating_count    = np.linspace(0, 100,  500)   # number of ratings [0 – 100]
x_platform        = np.linspace(1,  22,  500)   # platform count [1 – 22]
x_recent          = np.linspace(0,  10,  500)   # recency score [0 – 10]
x_recommendation  = np.linspace(0,  10,  500)   # output: recommendation [0 – 10]
```

```python
# ── 1c. Membership Function Sets ─────────────────────────────────────────

# Input 1: Global Score
mf_global = {
    'Poor'    : trimf(x_global, [0,   0,   2.5]),
    'Average' : trimf(x_global, [1.5, 2.5, 3.5]),
    'High'    : trimf(x_global, [3,   5,   5  ]),
}

# Input 2: Average User Rating
mf_avg_rating = {
    'Low'    : trimf(x_avg_user_rating, [1, 1, 3]),
    'Medium' : trimf(x_avg_user_rating, [2, 3, 4]),
    'High'   : trimf(x_avg_user_rating, [3, 5, 5]),
}

# Input 3: Rating Count
mf_count = {
    'Few'      : trimf(x_rating_count, [0,  0,   40 ]),
    'Moderate' : trimf(x_rating_count, [20, 50,  80 ]),
    'Many'     : trimf(x_rating_count, [60, 100, 100]),
}

# Input 4: Platform Count
mf_platform = {
    'Narrow' : trimf(x_platform, [1, 1,  6 ]),
    'Medium' : trimf(x_platform, [3, 7,  12]),
    'Wide'   : trimf(x_platform, [8, 22, 22]),
}

# Input 5: Recency
mf_recent = {
    'Old'    : trimf(x_recent, [0, 0,  4 ]),
    'Recent' : trimf(x_recent, [3, 5,  7 ]),
    'New'    : trimf(x_recent, [6, 10, 10]),
}

# Output: Recommendation Score
mf_recommendation = {
    'Poor'      : trimf(x_recommendation, [0, 0,  4 ]),
    'Fair'      : trimf(x_recommendation, [3, 5,  7 ]),
    'Excellent' : trimf(x_recommendation, [6, 10, 10]),
}
```

```python
# ── 1d. Fuzzify Helper ───────────────────────────────────────────────────
def fuzzify(global_in, avg_rating_in, count_in, platform_in, recent_in):
    """
    Convert crisp inputs into membership degrees for all linguistic variables.
    Returns a dict of dicts:  fuzzy_vals['global']['Poor'],  etc.
    """
    return {
        'global': {
            'Poor'    : float(trimf(global_in, [0,   0,   2.5])),
            'Average' : float(trimf(global_in, [1.5, 2.5, 3.5])),
            'High'    : float(trimf(global_in, [3,   5,   5  ])),
        },
        'avg_rating': {
            'Low'    : float(trimf(avg_rating_in, [1, 1, 3])),
            'Medium' : float(trimf(avg_rating_in, [2, 3, 4])),
            'High'   : float(trimf(avg_rating_in, [3, 5, 5])),
        },
        'count': {
            'Few'      : float(trimf(count_in, [0,  0,   40 ])),
            'Moderate' : float(trimf(count_in, [20, 50,  80 ])),
            'Many'     : float(trimf(count_in, [60, 100, 100])),
        },
        'platform': {
            'Narrow' : float(trimf(platform_in, [1, 1,  6 ])),
            'Medium' : float(trimf(platform_in, [3, 7,  12])),
            'Wide'   : float(trimf(platform_in, [8, 22, 22])),
        },
        'recent': {
            'Old'    : float(trimf(recent_in, [0, 0,  4 ])),
            'Recent' : float(trimf(recent_in, [3, 5,  7 ])),
            'New'    : float(trimf(recent_in, [6, 10, 10])),
        },
    }


# Quick sanity check
sample = fuzzify(4.5, 4.8, 85, 18, 9)
for var, degrees in sample.items():
    print(f'{var:12s}:', {k: round(v, 3) for k, v in degrees.items()})
```

## 2. Inference

Both Mamdani and Sugeno inference functions **receive the fuzzy values dict**
produced by `fuzzify()` above — no fuzzification is repeated here.

*(Rule base to be expanded by teammate.)*

```python
# ── 2a. Output Membership Shapes (used by Mamdani implication) ───────────
out_x         = np.linspace(0, 10, 500)
out_poor      = trimf(out_x, [0, 0,  4 ])
out_fair      = trimf(out_x, [3, 5,  7 ])
out_excellent = trimf(out_x, [6, 10, 10])
```

```python
# ── 2b. Mamdani Inference ────────────────────────────────────────────────
def mamdani_inference(fv):
    """
    Mamdani inference.
    fv  : fuzzy values dict returned by fuzzify()
    Returns the aggregated output array (shape: 500,) for defuzzification.
    """
    g = fv['global'];  r = fv['avg_rating']
    c = fv['count'];   p = fv['platform'];  rec = fv['recent']

    # Rule 1: IF global is Poor OR avg_rating is Low  →  Poor
    w1    = max(g['Poor'], r['Low'])
    rule1 = np.minimum(w1, out_poor)

    # Rule 2: IF global is Average AND count is Moderate  →  Fair
    w2    = min(g['Average'], c['Moderate'])
    rule2 = np.minimum(w2, out_fair)

    # Rule 3: IF global is High AND recent is New AND platform is Wide  →  Excellent
    w3    = min(g['High'], rec['New'], p['Wide'])
    rule3 = np.minimum(w3, out_excellent)

    # Aggregation (max of all clipped rule outputs)
    aggregated = np.maximum(rule1, np.maximum(rule2, rule3))
    return aggregated


# Quick test
fv_sample    = fuzzify(4.5, 4.8, 85, 18, 9)
mamdani_agg  = mamdani_inference(fv_sample)
print('Mamdani aggregated output (first 5 values):', mamdani_agg[:5].round(3))
```

```python
# ── 2c. Sugeno Inference ─────────────────────────────────────────────────
def sugeno_inference(fv):
    """
    Sugeno (Takagi-Sugeno) inference.
    fv  : fuzzy values dict returned by fuzzify()
    Returns a list of (firing_strength, crisp_output) tuples for defuzzification.
    """
    g = fv['global'];  r = fv['avg_rating']
    c = fv['count'];   p = fv['platform'];  rec = fv['recent']

    OUT_POOR, OUT_FAIR, OUT_EXCELLENT = 0.0, 5.0, 10.0

    # Rule 1: IF global is Poor OR avg_rating is Low  →  Poor
    w1 = max(g['Poor'], r['Low'])

    # Rule 2: IF global is Average AND count is Moderate  →  Fair
    w2 = min(g['Average'], c['Moderate'])

    # Rule 3: IF global is High AND recent is New AND platform is Wide  →  Excellent
    w3 = min(g['High'], rec['New'], p['Wide'])

    return [(w1, OUT_POOR), (w2, OUT_FAIR), (w3, OUT_EXCELLENT)]


# Quick test
sugeno_rules = sugeno_inference(fv_sample)
print('Sugeno rule outputs (weight, crisp_value):')
for i, (w, z) in enumerate(sugeno_rules, 1):
    print(f'  Rule {i}: weight={w:.3f}, output={z}')
```

```python
#Fuzzyfication
def trimf(x, a, b, c): 
    left = (x - a) / (b - a) if b != a else np.zeros_like(x)
    right = (x - c) / (b - c) if b != c else np.zeros_like(x)
    return np.maximum(0, np.minimum(left, right))
```

```python
#some very fucking rough mayor Mamdani 
def triangle(x, a, b, c):
    return np.maximum(0, np.minimum((x - a) / (b - a), (c - x) / (c - b)))

def mamdani_inference(input_val):
    # --- 1. Fuzzification ---
    # Input: Temperature
    is_cold = triangle(input_val, 0, 0, 50)
    is_hot  = triangle(input_val, 0, 50, 100) # Simple example

    # --- 2. Rule Evaluation (Implication) ---
    # Rule 1: IF Cold THEN Heater is HIGH
    # Rule 2: IF Hot THEN Heater is LOW
    
    # In Mamdani, we "clip" the output shape at the firing strength height.
    # Output shapes (Heater Power):
    x_universe = np.linspace(0, 100, 100) # Discrete points 0-100
    
    # Consequent shapes (Output fuzzy sets)
    heater_low  = triangle(x_universe, 0, 0, 50)
    heater_high = triangle(x_universe, 50, 100, 100)
    
    # Activation: Clip the output shapes
    # The output 'heater_high' is active to the degree that 'is_cold' is true
    rule1_output = np.minimum(is_cold, heater_high) 
    
    # The output 'heater_low' is active to the degree that 'is_hot' is true
    rule2_output = np.minimum(is_hot, heater_low)

    # --- 3. Aggregation ---
    # Combine all rules using MAX operator
    aggregated = np.maximum(rule1_output, rule2_output)
    
    # --- 4. Defuzzification (Centroid) ---
    # Formula: Sum(x * mu(x)) / Sum(mu(x))
    numerator = np.sum(x_universe * aggregated)
    denominator = np.sum(aggregated)
    
    if denominator == 0: return 0
    return numerator / denominator
```

```python
#some VERY rough fucking sugeno 
def trapezoid(x, a, b, c, d):
    """Manual membership function: Trapezoidal"""
    if x <= a or x >= d: return 0.0
    elif a < x < b: return (x - a) / (b - a)
    elif b <= x <= c: return 1.0
    elif c < x < d: return (d - x) / (d - c)
    return 0.0

def sugeno_inference(input_val):
    # --- 1. Fuzzification ---
    # Define antecedents (Inputs) manually
    # Example: Input is "Temperature" (0-100)
    cold = trapezoid(input_val, -10, 0, 20, 40)
    medium = trapezoid(input_val, 20, 40, 60, 80)
    hot = trapezoid(input_val, 60, 80, 100, 110)

    # --- 2. Rule Evaluation ---
    # Rule 1: IF Temp is Cold THEN Fan_Speed = 0 (Stop)
    # Rule 2: IF Temp is Medium THEN Fan_Speed = 50 (Slow)
    # Rule 3: IF Temp is Hot THEN Fan_Speed = 100 (Fast)
    
    # In Sugeno, the "consequent" is just a number (0, 50, 100)
    # The firing strength (weight) is the membership value itself.
    rules = [
        {'weight': cold, 'output': 0},
        {'weight': medium, 'output': 50},
        {'weight': hot, 'output': 100}
    ]
    
    # --- 3. Defuzzification (Weighted Average) ---
    numerator = sum(r['weight'] * r['output'] for r in rules)
    denominator = sum(r['weight'] for r in rules)
    
    if denominator == 0: return 0 # Avoid division by zero
    return numerator / denominator
```

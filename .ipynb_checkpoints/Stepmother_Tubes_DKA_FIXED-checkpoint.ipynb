## 1. Fuzzification

```python
def trimf(x, abc):
    """
    Triangular membership function (from scratch, no skfuzzy).
    abc : [a, b, c]  –  left foot, peak, right foot.
    """
    a, b, c = abc
    if a == b:                          # half-triangle: downward slope only
        return np.where(x < b, 0.0,
               np.where(x <= c, (c - x) / (c - b), 0.0))
    if b == c:                          # half-triangle: upward slope only
        return np.where(x < a, 0.0,
               np.where(x <= b, (x - a) / (b - a), 0.0))
    return np.maximum(0, np.minimum((x - a) / (b - a),
                                     (c - x) / (c - b)))
```

```python
# Universe of discourse for each linguistic variable
x_global          = np.linspace(0,   5,  500)
x_avg_user_rating = np.linspace(1,   5,  500)
x_rating_count    = np.linspace(0, 100,  500)
x_platform        = np.linspace(1,  22,  500)
x_recent          = np.linspace(0,  10,  500)
x_recommendation  = np.linspace(0,  10,  500)
```

```python
# Membership functions (used for visualisation / reference)
mf_global = {
    'Poor'    : trimf(x_global, [0,   0,   2.5]),
    'Average' : trimf(x_global, [1.5, 2.5, 3.5]),
    'High'    : trimf(x_global, [3,   5,   5  ]),
}
mf_avg_rating = {
    'Low'    : trimf(x_avg_user_rating, [1, 1, 3]),
    'Medium' : trimf(x_avg_user_rating, [2, 3, 4]),
    'High'   : trimf(x_avg_user_rating, [3, 5, 5]),
}
mf_count = {
    'Few'      : trimf(x_rating_count, [0,  0,   40 ]),
    'Moderate' : trimf(x_rating_count, [20, 50,  80 ]),
    'Many'     : trimf(x_rating_count, [60, 100, 100]),
}
mf_platform = {
    'Narrow' : trimf(x_platform, [1, 1,  6 ]),
    'Medium' : trimf(x_platform, [3, 7,  12]),
    'Wide'   : trimf(x_platform, [8, 22, 22]),
}
mf_recent = {
    'Old'    : trimf(x_recent, [0, 0,  4 ]),
    'Recent' : trimf(x_recent, [3, 5,  7 ]),
    'New'    : trimf(x_recent, [6, 10, 10]),
}
mf_recommendation = {
    'Poor'      : trimf(x_recommendation, [0, 0,  4 ]),
    'Fair'      : trimf(x_recommendation, [3, 5,  7 ]),
    'Excellent' : trimf(x_recommendation, [6, 10, 10]),
}
```

## 2. Inference

*(Rule base to be expanded by teammate.)*

```python
def mamdani_inference(global_in, avg_rating_in, count_in, platform_in, recent_in):
    """
    Mamdani fuzzy inference.
    Defuzzification: Centroid (Centre of Gravity) method.
    """
    # --- Fuzzification ---
    g_poor    = trimf(global_in, [0,   0,   2.5])
    g_average = trimf(global_in, [1.5, 2.5, 3.5])
    g_high    = trimf(global_in, [3,   5,   5  ])

    r_low    = trimf(avg_rating_in, [1, 1, 3])
    r_medium = trimf(avg_rating_in, [2, 3, 4])
    r_high   = trimf(avg_rating_in, [3, 5, 5])

    c_few      = trimf(count_in, [0,  0,   40 ])
    c_moderate = trimf(count_in, [20, 50,  80 ])
    c_many     = trimf(count_in, [60, 100, 100])

    p_narrow = trimf(platform_in, [1, 1,  6 ])
    p_medium = trimf(platform_in, [3, 7,  12])
    p_wide   = trimf(platform_in, [8, 22, 22])

    rec_old    = trimf(recent_in, [0, 0,  4 ])
    rec_recent = trimf(recent_in, [3, 5,  7 ])
    rec_new    = trimf(recent_in, [6, 10, 10])

    # --- Output membership shapes ---
    x_out         = np.linspace(0, 10, 500)
    out_poor      = trimf(x_out, [0, 0,  4 ])
    out_fair      = trimf(x_out, [3, 5,  7 ])
    out_excellent = trimf(x_out, [6, 10, 10])

    # --- Rule evaluation & implication ---
    # Rule 1: IF global is Poor OR avg_rating is Low  →  Poor
    w1    = np.maximum(g_poor, r_low)
    rule1 = np.minimum(w1, out_poor)

    # Rule 2: IF global is Average AND count is Moderate  →  Fair
    w2    = np.minimum(g_average, c_moderate)
    rule2 = np.minimum(w2, out_fair)

    # Rule 3: IF global is High AND recent is New AND platform is Wide  →  Excellent
    w3    = np.minimum(np.minimum(g_high, rec_new), p_wide)
    rule3 = np.minimum(w3, out_excellent)

    # --- Aggregation ---
    aggregated = np.maximum(rule1, np.maximum(rule2, rule3))

    # --- Defuzzification: Centroid ---
    denom = np.sum(aggregated)
    return (np.sum(x_out * aggregated) / denom) if denom != 0 else 5.0


mamdani_out = mamdani_inference(4.5, 4.8, 85, 18, 9)
print(f'Mamdani Recommendation Score: {mamdani_out:.2f} / 10')
```

```python
def sugeno_inference(global_in, avg_rating_in, count_in, platform_in, recent_in):
    """
    Sugeno (Takagi-Sugeno) fuzzy inference.
    Consequents are crisp constants; defuzzification uses weighted average.
    """
    # --- Fuzzification ---
    g_poor    = trimf(global_in, [0,   0,   2.5])
    g_average = trimf(global_in, [1.5, 2.5, 3.5])
    g_high    = trimf(global_in, [3,   5,   5  ])

    r_low    = trimf(avg_rating_in, [1, 1, 3])
    r_medium = trimf(avg_rating_in, [2, 3, 4])
    r_high   = trimf(avg_rating_in, [3, 5, 5])

    c_few      = trimf(count_in, [0,  0,   40 ])
    c_moderate = trimf(count_in, [20, 50,  80 ])
    c_many     = trimf(count_in, [60, 100, 100])

    p_narrow = trimf(platform_in, [1, 1,  6 ])
    p_medium = trimf(platform_in, [3, 7,  12])
    p_wide   = trimf(platform_in, [8, 22, 22])

    rec_old    = trimf(recent_in, [0, 0,  4 ])
    rec_recent = trimf(recent_in, [3, 5,  7 ])
    rec_new    = trimf(recent_in, [6, 10, 10])

    # --- Crisp consequent values ---
    OUT_POOR, OUT_FAIR, OUT_EXCELLENT = 0.0, 5.0, 10.0

    # --- Rule firing strengths ---
    # Rule 1: IF global is Poor OR avg_rating is Low  →  Poor
    w1 = np.maximum(g_poor, r_low)

    # Rule 2: IF global is Average AND count is Moderate  →  Fair
    w2 = np.minimum(g_average, c_moderate)

    # Rule 3: IF global is High AND recent is New AND platform is Wide  →  Excellent
    w3 = np.minimum(np.minimum(g_high, rec_new), p_wide)

    rules = [(w1, OUT_POOR), (w2, OUT_FAIR), (w3, OUT_EXCELLENT)]

    # --- Defuzzification: Weighted Average ---
    numer = sum(w * z for w, z in rules)
    denom = sum(w     for w, _ in rules)
    return (numer / denom) if denom != 0 else 5.0


sugeno_out = sugeno_inference(4.5, 4.8, 85, 18, 9)
print(f'Sugeno Recommendation Score:  {sugeno_out:.2f} / 10')
```
